### 5. New columns, Filter, Sorting data, Missing values and Handling missing values


```python
# Add a new columns
df['Price_per_sqft'] = df['Price_INR'] / df['Area_sqft']
print(df.head())
```

```python
# Filter data using condition
large_homes = df[df['Rooms'] > 5]
print(large_homes)
```

```python
# Sort the data
sorted_df = df.sort_values(by='Price_INR', ascending=False)
print(sorted_df.head())
```

```python
# Missing values
print(df.isnull().sum())
```

```python
# Handling missing values
df['Price_INR'] = df['Price_INR'].fillna(df['Price_INR'].mean())
```

## **[C] Data Visualization**

1.   Line plot
2.   Bar chart
3.   Histogram
4.   Scatter plot
5.   Box plot
6.   Pie chart



### 1. Line plot

```python
plt.figure(figsize=(8, 5))
plt.plot(df['Area_sqft'], df['Price_INR'], marker='o', linestyle='-', color='b')
plt.title('Property Valuation Trend')
plt.xlabel('Area (sqft)')
plt.ylabel('Price (INR)')
plt.grid(True)
plt.show()
```

### 2. Bar Chart

```python
avg_price = df.groupby('Rooms')['Price_INR'].mean()

plt.figure(figsize=(8, 5))
avg_price.plot(kind='bar', color='teal')
plt.title('Average Price by Room Count')
plt.xlabel('Number of Rooms')
plt.ylabel('Average Price (INR)')
plt.xticks(rotation=0)
plt.show()
```

### 3. Histogram

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Area_sqft'], bins=8, color='orange', edgecolor='black')
plt.title('Distribution of Property Areas')
plt.xlabel('Area (sqft)')
plt.ylabel('Number of Properties')
plt.show()
```

### 4. Scatter Plot

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Area_sqft'], df['Price_INR'], color='red', alpha=0.7)
plt.title('Area vs Price Correlation')
plt.xlabel('Area (sqft)')
plt.ylabel('Price (INR)')
plt.show()
```

### 5. Box Plot

```python
plt.figure(figsize=(8, 5))
plt.boxplot(df['Price_INR'], vert=False, patch_artist=True)
plt.title('Price Range & Market Spread')
plt.xlabel('Price (INR)')
plt.show()
```

### 6. Pie Chart

```python
room_counts = df['Rooms'].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(room_counts, labels=room_counts.index, autopct='%1.1f%%', startangle=140)
plt.title('Inventory Composition by Room Count')
plt.show()
```
